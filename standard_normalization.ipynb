```python
import numpy as np
```

```python
arr = np.array([[1,2],[3,4]])
```

```python
# Calculating mean
mean = np.mean(arr)
print(mean)
```

```python
# Calculating standard deviation
std_dev = np.std(arr)
print(std_dev)
```

```python
# Performing Standard Normalization
normalized_arr = (arr - mean)/std_dev
print(normalized_arr)
```

```python
print(np.mean(normalized_arr))
```
