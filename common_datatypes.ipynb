#### Common Datatypes

```python
import numpy as np
```

```python
# 1. int128,int64,int32,int16,int8
int_arr = np.array([10,20,30,40,50])
print(int_arr.dtype)
```

```python
# 2. float128,float,64,float32
float_arr = np.array([10.0,20.10,30.29,40.23])
print(float_arr.dtype)
```

```python
# 3. U7 (Unicode String)
str_arr = np.array(["hi","hello","welcome","bye"])
print(str_arr.dtype)
```

```python
# 4. complex64,complex128
complex_arr = np.array([3+2j,6+9j])
print(complex_arr.dtype)
```

```python
# 5. bool
bool_arr = np.array([True,True,False])
print(bool_arr.dtype)
```

```python
# 6. object
object_arr = np.array([1,2,3,{10,20,30},"Hi",True])
print(object_arr.dtype)
```
