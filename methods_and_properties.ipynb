```python
import numpy as np
```

```python
#From python list
arr = np.array([1,2,3,4,5])
print(arr)
print(type(arr))
print(arr.dtype)

arr = np.array([1,2,3,4,"Hello"])
print(arr)
print(type(arr))
print(arr.dtype) # returns type of value present inside the np array
print(arr.shape) # returns the dimensions of an np array (tuple)
```

### Creating the Numpy Array from scratch using np methods

```python
# prefill the array with zeros
# zero(tuple->shape) methdo
zero_arr = np.zeros((2,3),dtype="int64")
print(zero_arr)
print(zero_arr.shape)
print(zero_arr.dtype)
```

```python
# prefill the array with ones
# ones(tuple->shape) methdo
ones_arr = np.ones((2,),dtype="int64") # 1d array
print(ones_arr)
print(ones_arr.shape)
print(ones_arr.dtype)
```

```python
# prefill the array with perticular value
# full(shape,value) method
any_arr = np.full((4,4),100)
print(any_arr)
```

```python
# creating the identity matrix
# eye(number->square matrix)
eye_arr = np.eye(3,dtype="int64") 
print(eye_arr)
```

```python
# creating a 1D array between the range of values
# aranage(start,end,step)
range_arr = np.arange(0,11,2) # even numbers array
print(range_arr)
```

```python
# creating the evenly spaced arrays
# linspace(start,end,size)
lin_arr = np.linspace(1,100,3,dtype="int64")
print(lin_arr)
```

### Most important numpy array properties

```python
# 1.shape -> Retunrs the dimensions of an array in the form of tuple (row,column)
arr = np.array([1,2,3,4,5])
print(arr.shape) # (5,)

arr2d = np.array([[1,2,3],[4,5,6]])
print(arr2d.shape) # (2,3)
```

```python
# 2.size -> Retunrs the total no.of elements present inside an array
print(arr.size) # 5
```

```python
# 3.dtype -> Returns the data type of the value present inside the numpy array
print(arr.dtype) # int64 (or) int32
```

```python
# 4.ndim -> Returns the dimension in the integer number
print(arr2d.ndim) # 2
```

#### Converting one type array to another type

astype() method will always retunrs the new array

```python
# np_array.astype(np.dtype_name)
arr = np.array([1,2,3,4,5]) # int64
print(arr)
float_arr = arr.astype(np.float64) # returns a new float64 array
print(float_arr)
print(float_arr.dtype)
```
