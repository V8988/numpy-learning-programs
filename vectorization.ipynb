### Vectorization

```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
```

```python
# Squares of each value presnt inside the array
print(arr ** 2)
```

```python
# Adding 10 for each value in the array
print(arr + 10)
```

```python
# Adding two differnet array
arr2 = np.array([6,7,8,9,10])
print(arr+arr2)
```
