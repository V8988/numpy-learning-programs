### Generate a random number

```python
from numpy import random as rd
```

```python
# randint(lower,upper) -> upper value is exclusive
print(rd.randint(1,3)) 
```

```python
def gen_otp():
    count = 1
    while(count<=4):
        print(rd.randint(0,10),end="\t")
        count+=1
```

```python
gen_otp()
```
