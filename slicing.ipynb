```python
import numpy as np
```

### Slicing the 1D Array

While slicing the numpay array it returns the view not a copy like python list

view : A sub array from the original array with out creating any extra copy

```python
arr = np.array([1,2,3,4,5,6])
# numpy_array[start_index:end_index,step_value]
# Here ending index value is exclusive
print(arr[0:1])
print(arr[0:2])
print(arr[2:4])
```

```python
# Copy
py_list = [1,2,3,4,5]
sub_list = py_list[1:3]
sub_list[0] = 100
print(py_list) # original list will not effected
print(sub_list)
```

```python
# View
np_arr = np.array([1,2,3,4,5])
sub_arr = np_arr[0:2]
sub_arr[0] = 100
print(np_arr) # original array will be effected
print(sub_arr)
```

#### To get the copy of slice(sub array) of numpy array use .copy() method

```python
np_arr = np.array([1,2,3,4,5])
sub_arr = np_arr[0:2].copy()
sub_arr[0] = 100
print(np_arr) # original array will not be effected
print(sub_arr)
```
