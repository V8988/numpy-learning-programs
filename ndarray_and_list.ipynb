## Performance Comparision Test (List V/S NDarray)

```python
import numpy as np
import time
```

```python
# execution performance
size = 100_000000

# python list
py_list = list(range(size))
start_time = time.time()
sq_list = [x**2 for x in py_list]
end_time= time.time()
print(f"Python List Time : {end_time - start_time} sec.")

# np ndarray
np_arr = np.array(py_list)
start_time = time.time()
sq_array = np_arr ** 2 # Vectorization
end_time = time.time()
print(f"Numpy NDArray Time : {end_time - start_time} sec.")
```
