### Math Methods

```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
```

#### Aggreate Functions

```python
# 1.sum()
print(np.sum(arr))
```

```python
# 2.prod() 
print(np.prod(arr))
```

```python
# 3.min()
print(np.min(arr))
```

```python
# 4.max()
print(np.max(arr))
```

```python
# 5.argmin() -> Retunrs the index value of min value
print(np.argmin(arr))
```

```python
# 6.argmax() -> Retunrs the index value of max value
print(np.argmax(arr))
```

```python
# 7.mean()
print(np.mean(arr))
```

```python
# 8.std() -> Retunrs the standard deviation of arr values
print(np.std(arr))
```

```python
# 9.median()
print(np.median(arr))
```

```python
# 10.var() -> Retunrs the variance of arr values
print(np.var(arr))
```

#### Power Funcstion

```python
# 1.square(number/arr)
print(np.square(2))
```

```python
# 2.sqrt(number/arr)
print(np.sqrt(2))
```

```python
# 3.pow(number(a)/arr,b)
print(np.pow(2,4))
```

#### Log and Exponential

```python
# 1.log(number/arr)
print(np.log(arr))
```

```python
# 2.log2(number/arr)
print(np.log2(arr))
```

```python
# 3.exp(number/arr)
print(np.exp(arr))
```

#### Rounding Methods

```python
# 1.round(number/arr)
print(np.round(3.14))
```

```python
# 2.ceil(number/arr)
print(np.ceil(3.14))
```

```python
# 3.floor(number/arr)
print(np.floor(3.95))
```

```python
# 4.trunc(number/arr)
print(np.trunc(3.87)) # replace the decimal part with zero
```

#### Extra Math Methdos

```python
# 1.unique()
print(np.unique([1,2,3,4,2,3,4,2,4,2]))
```

```python
# 2.sort()
print(np.sort([8,4,5,1,5,0,2]))
```

```python
# 3.abs()
print(np.abs([1,-2,10,-29,76,49,-65])) # removes the negitive sign
```
